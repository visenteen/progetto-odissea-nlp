```python
# CELLA 2: Caricamento in memoria dei file reali e conteggio delle parole

# 1. Leggiamo il file dell'Odissea di Omero in inglese
with open("omero.txt", "r", encoding="utf-8") as f1:
    testo_omero = f1.read()

# 2. Leggiamo il file della sceneggiatura di Nolan
with open("nolan.txt", "r", encoding="utf-8") as f2:
    sceneggiatura_nolan = f2.read()

# 3. Calcoliamo il numero totale di parole per la statistica descrittiva
parole_omero = len(testo_omero.split())
parole_nolan = len(sceneggiatura_nolan.split())

print("=" * 60)
print("FILE CARICATI CON SUCCESSO NELLA MEMORIA DI PYTHON!")
print("=" * 60)
print(f"L'Odissea di Omero contiene: {parole_omero} parole.")
print(f"La sceneggiatura di Nolan contiene: {parole_nolan} parole.")
print("=" * 60)
print("Ora puoi passare alla Cella 3 per il calcolo semantico.")
```

```python
# CELLA 3: Calcolo della Somiglianza Semantica tramite Embeddings

import spacy

# 1. Carichiamo la "mente" di SpaCy in inglese
nlp = spacy.load("en_core_web_lg")

# 2. Estraiamo una porzione significativa di testo per non sovraccaricare la memoria cloud
# (I primi 50.000 caratteri sono un campione perfetto dal punto di vista accademico)
doc_omero = nlp(testo_omero[:50000])
doc_nolan = nlp(sceneggiatura_nolan[:50000])

# 3. L'algoritmo confronta i vettori di significato (Cosine Similarity tra Embeddings)
punteggio_finale = doc_omero.similarity(doc_nolan)

print("=" * 60)
print("  PROGETTO: UN MARE DI DATI O UNA CONNESSIONE UMANA?  ")
print("=" * 60)
print(f"Indice di Somiglianza Semantica (AI Level): {punteggio_finale:.4f}")
print("=" * 60)
print("Nota: Il punteggio varia da 0.0000 (opposti) a 1.0000 (identici).")
```

```python
# CELLA 4: Analisi del Contesto - Influenza della Filmografia di Nolan

# 1. Definiamo i riassunti dei film chiave di Nolan (il contesto)
# Nota: puoi espandere questi testi inserendo trame più lunghe prese da Wikipedia!
contesto_filmografia = {
    "Interstellar": "A journey through wormholes and space-time relativity to save humanity, focusing on the deep emotional bond and separation between a father and his children across cosmic distances.",
    "Inception": "A complex heist within layers of human dreams and subconscious memory, where the protagonist is haunted by the guilt of his past and fights endlessly to return home to his children.",
    "Tenet": "A secret agent manipulates the flow of time and entropy to prevent a global apocalypse, dealing with fatalism, destiny, and highly complex physical laws."
}

print("=" * 60)
print("  ANALISI DI CONTESTO: QUALE FILM INFLUENZA DI PIÙ ODYSSEY?  ")
print("=" * 60)

# 2. Utilizziamo il modello SpaCy in inglese (nlp) già caricato per confrontare i contesti
# Confrontiamo ciascun film con i primi 50.000 caratteri della sceneggiatura (doc_nolan)
for titolo_film, riassunto in contesto_filmografia.items():
    doc_contesto = nlp(riassunto)

    # Calcolo della somiglianza semantica tramite Embeddings
    somiglianza_contesto = doc_nolan.similarity(doc_contesto)

    print(f"Somiglianza con {titolo_film}: {somiglianza_contesto:.4f}")

print("=" * 60)
```
